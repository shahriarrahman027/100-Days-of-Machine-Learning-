```python
titanic=pd.read_csv("titanic.csv")
```

```python
titanic.head()
```

# Numerical-Categorical

```python
sns.barplot(y=titanic["survived"],x=titanic["class"],hue=titanic["sex"])
```

```python
sns.boxplot(x=titanic["class"],y=titanic["age"])
```

```python
sns.boxplot(x=titanic["class"],y=titanic["age"],hue=titanic["survived"])
```

```python
sns.kdeplot(data=titanic,x="age", hue="survived")
```

# Categorical - Categorical 

```python
(titanic.groupby('class')['survived'].mean()*100).plot(kind='bar')
```

```python
sns.heatmap(pd.crosstab(titanic['class'],titanic['survived']))
```

```python
sns.clustermap(pd.crosstab(titanic['class'],titanic['survived']))
```

```python
sns.pairplot(titanic,hue='survived')
```

```python
sns.pairplot(tips,hue='tip')
```

```python
sns.lineplot(x=titanic['age'],y=titanic['class'])
```
