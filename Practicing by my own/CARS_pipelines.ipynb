```python
import numpy as np 
import pandas as pd 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.tree import DecisionTreeClassifier 
```

```python
df=pd.read_csv("cars.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.drop(["selling_price"],axis=1),
                                                       df["selling_price"],
                                                       test_size=.3,
                                                       random_state=42)
```

```python
X_train.head()
```

# ordinal encoding

```python
trf1=ColumnTransformer([
    ('owner',OrdinalEncoder(),[3])
],remainder="passthrough")
```

```python
X_train.head()
```

```python
# One Hot encoding 
```

```python
trf2=ColumnTransformer([
    ('brand_fuel',OneHotEncoder(sparse_output=False,handle_unknown="ignore"),['brand','fuel'])
     ],remainder="passthrough")
```

```python
# Scaling 
```

```python
trf3=ColumnTransformer([
    ('scaled',MinMaxScaler(),slice(0,8))
])
```

```python
trf4=SelectKBest(score_func=chi2,k=5)
```

```python
trf5=DecisionTreeClassifier()
```

```python
pipe=Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
    ('trf3',trf3),
    ('trf4',trf4),
    ('trf5',trf5)
])
```

```python
pipe.fit(X_train,y_train)
```
