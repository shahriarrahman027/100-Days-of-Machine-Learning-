```python
import pandas as pd
import seaborn as sns 
```

```python
df=pd.read_csv("cars.csv")
df.head()
```

# Categorical Data

```python
sns.countplot(x='brand',data=df)
```

```python
sns.countplot(x='fuel',data=df)
```

```python
df['brand'].value_counts().plot(kind='pie',autopct='%2f')
```

```python
df['owner'].value_counts().plot(kind='pie',autopct='%2f')
```
