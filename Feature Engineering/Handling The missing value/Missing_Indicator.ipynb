```python
import numpy as np 
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split 
from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
df=pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.drop(['Survived'],axis=1),
                                              df['Survived'],
                                              test_size=.3,
                                              random_state=42)
X_train.shape,y_train.shape
```

```python
# check the accuracy without missing indicator 
```

```python
si=SimpleImputer()
X_train_trf=si.fit_transform(X_train)
X_test_trf=si.transform(X_test)
X_train_trf
```

```python
from sklearn.linear_model import LogisticRegression 
clf=LogisticRegression()
clf.fit(X_train_trf,y_train)
y_pred=clf.predict(X_test_trf)
from sklearn.metrics import accuracy_score 
accuracy_score(y_test,y_pred)
```

```python
# check the accuracy score using the missing  indicator 
```

```python
mi=MissingIndicator()
X_train_missing=mi.fit_transform(X_train)
X_test_missing=mi.transform(X_test)
```

```python
X_test_missing
```

```python
X_train[['Age_NA','Fare_NA']]=X_train_missing
X_train
```

```python
X_test[['Age_NA','Fare_NA']]=X_test_missing
X_test
```

```python
si=SimpleImputer()
X_train_trf2=si.fit_transform(X_train)
X_test_trf2=si.transform(X_test)
```

```python
clf=LogisticRegression()
clf.fit(X_train_trf2,y_train)
y_pred=clf.predict(X_test_trf2)
accuracy_score(y_test,y_pred)
```
