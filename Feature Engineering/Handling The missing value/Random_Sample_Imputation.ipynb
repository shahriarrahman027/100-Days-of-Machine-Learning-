```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv('titanic_toy.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
df.sample(15)
```

```python
#df['Age_imputed']=df['Age'].fillna(df['Age'].dropna().sample(1).iloc[0])

df['Age_imputed'] = df['Age'].copy()

missing = df['Age'].isna()

df.loc[missing, 'Age_imputed'] = np.random.choice(
    df['Age'].dropna(),
    size=missing.sum()
)
```

```python
df.head(20)
```

```python
sns.kdeplot(df['Age'],label='orginal')
sns.kdeplot(df['Age_imputed'],label='imputed')
```
