```python
import pandas as pd
import seaborn as sns
import scipy
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic=pd.read_csv("Titanic.csv")
titanic.head()
```

# Scatterplot(Numerical - Numerical) 

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
```

## men women speration 

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'])
```

## categorized by smoking / non smoking 

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'])
```

# Barplot ( Numerical - Categorical )

```python
titanic.head()
```

```python
sns.barplot(x=titanic['pclass'],y=titanic['age'])
```

## for more camparison ( multivariate analysis)

```python
sns.barplot(x=titanic['pclass'],y=titanic['age'],hue=titanic['sex'])
```

## Boxplot

```python
sns.boxplot(x=titanic['sex'],y=titanic['age'])
```

## boxplot(multi)

```python
sns.boxplot(x=titanic['sex'],y=titanic['age'],hue=titanic['survived'])
```

# Displot( Numerical - categorical ) 

```python
sns.distplot(titanic[titanic['survived']==0]['age'],hist=False)
sns.distplot(titanic[titanic['survived']==1]['age'],hist=False)
```

# HeatMap( categorical - categorical)

```python
sns.heatmap(pd.crosstab(titanic['pclass'],titanic['survived']))
```

## showing perchentage 

```python
(titanic.groupby('pclass')['survived'].mean()*100).plot(kind='bar')
```

# Clustermap(categorical-categorical)

```python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['survived']))
```
